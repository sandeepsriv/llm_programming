```python
import tiktoken
```

```python
encoding = tiktoken.encoding_for_model("gpt-4.1-mini")

# word count = 6
# word count with punctuation = 8

token = encoding.encode("Hello, my name is Sandeep Srivastava.")
```

```python
token
```

```python
for token_id in token:
    token_text = encoding.decode([token_id])
    print(f"Token ID {token_id} = {token_text}")
```
