```python
from langchain_openai import ChatOpenAI
from IPython.display import Markdown, display
```

```python
explain_convexity = [
    {"role": "user", "content": "What is convexity in finance. Explain in simple terms and how does it compare to refrence of convexity made by Nassim Taleb in his books"},
]
```

```python

llm = ChatOpenAI(model="gpt-5-mini")
response = llm.invoke(explain_convexity)
```

```python
display(Markdown(response.content))
```
