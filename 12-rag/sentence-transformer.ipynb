```python
from sentence_transformers import SentenceTransformer
from langchain_huggingface import HuggingFaceEmbeddings
from dotenv import load_dotenv
import os
```

```python
load_dotenv(override=True)

OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")
HUGGINGFACE_ACCESS_TOKEN = os.getenv("HUGGINGFACE_ACCESS_TOKEN")

if OPENAI_API_KEY:
    print(f"OpenAI API Key exists and begins {OPENAI_API_KEY[:8]}")
else:
    print("OpenAI API Key not set")


if HUGGINGFACE_ACCESS_TOKEN:
    print(f"HUGGINGFACE_ACCESS_TOKEN API Key exists and begins {HUGGINGFACE_ACCESS_TOKEN[:8]}")
else:
    print("HUGGINGFACE_ACCESS_TOKEN API Key not set")
```

```python
model = SentenceTransformer("all-MiniLM-L6-v2")
# embeddings = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")
```

```python
sentences = [
    "The dog sits outside waiting for a teat.",
    "I am going swimming",
    "I am fan of Mohd. Rafi",
    "I am listening to Shammi Kapoor movie songs"
] 
```

```python
embeddings = model.encode(sentences)
print(embeddings.shape)
```

```python

similarities = model.similarity(embeddings,embeddings)
print(similarities)
```
