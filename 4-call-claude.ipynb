```python
from anthropic import Anthropic
```

```python
anthropic = Anthropic()

response = anthropic.messages.create(
    model="claude-sonnet-4-5",
    messages=[
        {
            "role": "user",
            "content" : "Describe python programming language yield operator"
        }
    ],
    max_tokens=100
)
```

```python
print(response.content[0].text)
```
